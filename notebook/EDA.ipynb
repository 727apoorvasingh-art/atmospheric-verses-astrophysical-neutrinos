```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/simulated_neutrino_events.csv')
print(df.shape)
df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df[df['label']==0]['energy'], bins=50, alpha=0.6, label='Atmospheric (0)')
plt.hist(df[df['label']==1]['energy'], bins=50, alpha=0.6, label='Astrophsical (1)')
plt.xscale('log')
plt.xlabel('Energy (Gev)')
plt.ylabel('count')
plt.title('Energy distribution by class')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df[df['label']==0]['zenith'], bins=50, alpha=0.6, label='Atmospheric (0)')
plt.hist(df[df['label']==1]['zenith'], bins=50, alpha=0.6, label='Astrophsical (1)')
plt.xlabel('zenith')
plt.ylabel('count')
plt.title('zenith distribution by class')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df[df['label']==0]['energy'], df[df['label']==0]['zenith'], alpha=0.3, s=10, label='Atmospheric (0)')
plt.scatter(df[df['label']==1]['energy'], df[df['label']==1]['zenith'], alpha=0.3, s=10, label='Astrophsical (1)')
plt.xscale('log')
plt.xlabel('Energy (Gev)')
plt.ylabel('zenith')
plt.title('Energy vs zenith distribution by class')
plt.legend()
plt.show()
```
